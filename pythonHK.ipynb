```python
# 두 숫자 입력
num1 = float(input("첫 번째 숫자를 입력하세요: "))
num2 = float(input("두 번째 숫자를 입력하세요: "))

# 연산자 입력
operator = input("연산자를 입력하세요 (+, -, *, /): ")

# 계산 및 결과 출력
if operator == "+":
    result = num1 + num2
elif operator == "-":
    result = num1 - num2
elif operator == "*":
    result = num1 * num2
elif operator == "/":
    if num2 != 0:
        result = num1 / num2
    else:
        result = "0으로 나눌 수 없습니다."
else:
    result = "올바른 연산자를 입력해주세요."

print("결과:", result)
```

```python
slist = [10, 20, 30, 40, 50] #여러 개의 값들을 묶은 것, 순번 0번 부터 시작
print(slist)

slist.append("사과") #값을 리스트 뒤에 추가
print(slist)

slist.insert(1,"스파게티") #값을 인덱스 1에 추가
print(slist)

slist.remove(10) #10 이라는 값 찾아서 삭제
print(slist)

slist.pop(1) #인덱스 1번 값 삭제(위치기반)
print(slist)
```
